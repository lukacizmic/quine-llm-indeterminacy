```python
# Find 50 concepts from the dataset each having higher naming consistency, start from the concept with the lowest consistency in the dataset and end with the highest consistency,
# make it so that each concept has a proportionately higher naming consistency than the previous one, ending with the concept having the highest consistency,
# do not use entries with the same naming consistency, each selected concept should have a unique naming consistency value
# Nan values are excluded from the selection
# Since each word has several rows which represent different images presented to participants, we need to make sure that naming consistency is 
# calculated as the average naming consistency across all images for each word

def get_top_50_concepts(things_dataset):
    
    things_dataset = things_dataset.copy()

    things_dataset = things_dataset.dropna(subset=["nameability_naming-consistency"])

    things_dataset = things_dataset.groupby("Word", as_index=False)["nameability_naming-consistency"].mean()

    sorted_things = things_dataset.sort_values(by="nameability_naming-consistency")
    top_50_concepts = sorted_things.iloc[::len(sorted_things)//50, :].head(50)
    
    return top_50_concepts

get_top_50_concepts(things_dataset)
```

```python
list_of_top_50_concepts = get_top_50_concepts(things_dataset)["Word"].tolist()

with open("C:\\Users\\Luka\\Documents\\Projects\\quine-llm\\list_of_top_50_concepts.txt", "w") as f:
    for concept in list_of_top_50_concepts:
        f.write(f"{concept}\n")
        
# What does the above code do?

# The above code retrieves the top 50 concepts from the dataset based on naming consistency, stores them in a list, and writes them to a text file.

# Where to find the txt file
# The text file "list_of_top_50_concepts.txt" will be saved in "C:\Users\Luka\Documents\Projects\quine-llm\".

# Can't find the text file? Make sure to check the current working directory.

#
```
